```python
# Importação de bibliotecas externas necessárias
import sys
import os
import cProfile as cpr
from pyprof2calltree import convert

sys.path.append(os.path.abspath("../..")) 


import logging
from Modules.Models import * 
from Modules.Methods import *
from Modules.Handlers import *
from datetime import datetime



string_data = datetime.now().strftime("%d_%m_%Y_%H_%M")



```

```python
gen=10
handler = ExecutionHandler(
    model=ModelWrapper.GRN5(train_percentage=1.0),              #"configures" the "model-object", to be ready to use
    method_class=DE,
    seeds=[0],
    solvers=['ODEINT'],
    errors=['MSE'],
    generations=gen,
    parallel=False,
)

handler
```

```python
if __name__ == "__main__":    #garantees that the code will be profilled when ran, not imported
    with cpr.Profile() as profile: #runs everything inside the block using the Profile()
        handler.execute()

    callgrind_filename = f"callgrind.{string_data}_GRN5_DE_Parallel_{gen}_generations_results.out"
    convert(profile.getstats(), callgrind_filename)
    print(f"Profiling complete! Open QCacheGrind and load: {callgrind_filename}")
```
