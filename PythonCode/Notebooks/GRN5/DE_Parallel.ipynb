```python
# Importação de bibliotecas externas necessárias
import sys
import os
import cProfile as cpr
import pstats as pStats


sys.path.append(os.path.abspath("../..")) 


import logging
from Modules.Models import * 
from Modules.Methods import *
from Modules.Handlers import *
```

```python
handler = ExecutionHandler(
    model=ModelWrapper.GRN5(train_percentage=0.6),
    method_class=DE,
    seeds=[0, 1],
    solvers=['ODEINT'],
    errors=['MSE'],
    generations=1,
    parallel=False,
)

handler
```

```python
if __name__ == "__main__":    #garantees that the code will be profilled when ran, not imported
    with cpr.Profile() as profile: #runs everything inside the block using the Profile()
        handler.execute()
    results = pStats.Stats(profile) #creates stats object with profile data
    results.sort_stats(pStats.SortKey.TIME)
    results.dump_stats("results.prof") #dumps results to a file
```
